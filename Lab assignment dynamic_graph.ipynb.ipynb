```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
pio.templates.default = "plotly_white"
print("plotly version:" , __import__('plotly').__version__)
```

```python
df = pd.read_csv("cleaned_india_city_climate_expanded.csv")
```

```python
print("shape:" , df.shape)
df.head
```

```python
fig1 = px.histogram(
    df,
    x="tavg",
    title="Distribution of Average Temperature (India)",
    labels={"tavg": "Average Temperature (°C)"}
)

fig1.update_traces(
    xbins=dict(
        start=-10,
        end=50,
        size=2
    )
)

fig1.update_layout(
    yaxis_title="Number of Records",
    bargap=0.05,
    xaxis=dict(
        rangeslider=dict(visible=True, thickness=0.12),
        dtick=5
    )
)

fig1.show()
```

```python
fig = px.scatter(
    df,
    x="tmin",
    y="tmax",
    color="season",
    hover_data=["city", "year", "month", "tavg"],
    title="Minimum vs Maximum Temperature by Season",
    labels={
        "tmin": "Minimum Temperature (°C)",
        "tmax": "Maximum Temperature (°C)",
        "season": "Season"
    },
    opacity=0.75
)

fig.update_layout(
    template="plotly_white",
    height=600,
    hovermode="closest",
    legend_title="Season"
)

fig.update_traces(marker=dict(size=9))

fig.show()
```

```python
fig = px.bar(
    df,
    x="city",
    y="tavg",
    color="season",
    hover_data=["year", "month", "tmin", "tmax"],
    title="Average Temperature by City and Season",
    labels={
        "city": "City",
        "tavg": "Average Temperature (°C)",
        "season": "Season"
    },
    barmode="group"
)

fig.update_layout(
    template="plotly_white",
    height=600,
    xaxis_tickangle=-45,
    yaxis_title="Average Temperature (°C)",
    legend_title="Season"
)

fig.show()
```

```python
fig = px.pie(
    df,
    names="season",
    title="Distribution of Climate Records by Season",
    hover_data=["city"],
    hole=0
)

fig.update_layout(
    template="plotly_white",
    height=600,
    legend_title="Season"
)

fig.update_traces(
    textinfo="percent+label",
    textposition="inside"
)

fig.show()
```
